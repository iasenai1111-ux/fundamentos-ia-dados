## Parte 4 — Prompt vago × prompt PACIF ✍️
Cole abaixo os dois pedidos que você fez para a IA (Gemini ou ChatGPT).

**Um bom prompt tem 5 partes (PACIF):**
1. **Persona** — quem a IA deve ser ("Atue como comprador de uma construtora")
2. **Ação** — o que fazer ("escreva um e-mail ao fornecedor")
3. **Contexto** — a situação ("o cimento atrasou 3 dias e a obra parou")
4. **Intenção** — para que serve ("conseguir nova data até amanhã às 12h")
5. **Formato** — como entregar ("tom educado e firme, no máximo 6 linhas")

> 🔒 **Nunca** coloque na IA: CPF, senhas, salários, dados de clientes ou documentos internos da empresa.

```python
prompt_fraco = ""

prompt_bom = ""

# O que mudou na resposta da IA quando você usou o prompt PACIF?
o_que_mudou = ""

partes = ["Persona", "Ação", "Contexto", "Intenção", "Formato"]
print("Confira se o seu prompt PACIF tem as 5 partes:")
for parte in partes:
    print("  ☐", parte)

if prompt_bom.strip() == "":
    print("\n⬜ Cole o seu prompt bom entre as aspas.")
elif len(prompt_bom) > len(prompt_fraco):
    print(f"\n✅ Seu prompt bom tem {len(prompt_bom)} letras; o fraco tinha {len(prompt_fraco)}. Mais contexto, resposta melhor.")
else:
    print("\n🤔 Seu prompt bom está mais curto que o fraco. Faltou contexto?")
_PROGRESSO["Parte 4"] = len(prompt_bom) > 0
```
